```python
import scanpy as sc
import pandas as pd
from scipy.io import mmread

X = mmread("matrix.mtx").T.tocsr()
obs = pd.read_csv("metadata.csv", index_col=0)
var = pd.DataFrame(index=pd.read_csv("genes.csv", index_col=0).iloc[:, 0])

adata = sc.AnnData(X=X, obs=obs, var=var)
adata.obsm["X_pca"] = pd.read_csv("pca.csv", index_col=0).values
adata.obsm["X_harmony"] = pd.read_csv("harmony.csv", index_col=0).values
adata.obsm["X_umap"] = pd.read_csv("umap.csv", index_col=0).values

adata.write("hnc_myeloid_2021.h5ad")
```
