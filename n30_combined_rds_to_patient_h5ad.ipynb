```python
import rpy2.robjects as ro
from rpy2.robjects import r
from pathlib import Path
import gzip, shutil
```

```python
src = Path("data/combined_n30_anal_annotated.rds")   # gzipped content
dst = src.with_suffix(".rds.raw")                   # decompressed output

print("Decompressing to:", dst)
with gzip.open(src, "rb") as f_in, open(dst, "wb") as f_out:
    shutil.copyfileobj(f_in, f_out, length=1024*1024*16)  # 16MB chunks

print("done. raw size bytes:", dst.stat().st_size)
```

```python
raw_path = "data/combined_n30_anal_annotated.rds.raw"

# 1) Try readRDS
try:
    obj = r['readRDS'](raw_path)
    print("readRDS worked.")
    print("class:", r['class'](obj))
except Exception as e:
    print("readRDS failed:", e)

    # 2) Try load() in case it's an .RData-like file
    loaded = r['load'](raw_path)
    loaded_names = list(loaded)
    print("load() worked. Objects:", loaded_names)

    obj = r[loaded_names[0]]
    print("Using first object:", loaded_names[0])
    print("class:", r['class'](obj))
```
