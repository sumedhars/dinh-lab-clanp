```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
def compute_within_k_gep_correlation(gene_spectra_file, show_plot=True):
    """
    Compute within-K GEP×GEP correlation matrix from gene spectra file.
    
    Parameters:
    -----------
    gene_spectra_file : str
        Path to the gene spectra file (TSV format)
    show_plot : bool
        Whether to display the correlation heatmap (default: True)
    
    Returns:
    --------
    correlation_matrix : pd.DataFrame
        Correlation matrix with GEP indices as index and columns
    """
    # Read the gene spectra file (GEPs as rows, genes as columns)
    gene_spectra = pd.read_csv(gene_spectra_file, sep='\t', index_col=0)
    
    # Remove any columns with zero variance (constant values)
    non_constant_cols = gene_spectra.std() > 0
    gene_spectra_filtered = gene_spectra.loc[:, non_constant_cols]
    
    # Compute correlation matrix between GEPs (rows)
    correlation_matrix = gene_spectra_filtered.T.corr(method='pearson')
    
    # Display heatmap if requested
    if show_plot:
        plt.figure(figsize=(10, 8))
        sns.heatmap(correlation_matrix, 
                    annot=True,  # Show correlation values
                    fmt='.2f',   # Format to 2 decimal places
                    cmap='coolwarm',  # Color scheme
                    center=0,    # Center colormap at 0
                    vmin=-1, vmax=1,  # Correlation range
                    square=True,  # Make cells square
                    cbar_kws={'label': 'Pearson Correlation'})
        plt.title('Within-K GEP×GEP Correlation Matrix', fontsize=14, pad=20)
        plt.xlabel('GEP', fontsize=12)
        plt.ylabel('GEP', fontsize=12)
        plt.tight_layout()
        plt.show()
    
    return correlation_matrix
```

```python
def plot_gep_comparisons(df1, df2, gep_pairs, figsize=(15, 10), ncols=3, output_dir='gep_comparisons'):
    """
    Create multiple scatterplots comparing GEPs from two dataframes and save residuals to CSV.
    
    Parameters:
    -----------
    df1 : pandas.DataFrame
        First dataframe where rows are GEPs and columns are genes
    df2 : pandas.DataFrame
        Second dataframe where rows are GEPs and columns are genes
    gep_pairs : list of tuples
        List of (df1_row_idx, df2_row_idx) pairs to compare
        Example: [(0, 0), (1, 1), (0, 1), (2, 3)]
    figsize : tuple
        Figure size (width, height)
    ncols : int
        Number of columns in the subplot grid
    output_dir : str
        Directory name where CSV files will be saved
        
    Returns:
    --------
    fig, axes : matplotlib figure and axes objects
    """
    import os
    import pandas as pd
    
    # Create output directory if it doesn't exist
    os.makedirs(output_dir, exist_ok=True)
    
    n_plots = len(gep_pairs)
    nrows = int(np.ceil(n_plots / ncols))
    fig, axes = plt.subplots(nrows, ncols, figsize=figsize)
    axes = axes.flatten() if n_plots > 1 else [axes]
    
    for idx, (gep1_idx, gep2_idx) in enumerate(gep_pairs):
        ax = axes[idx]
        
        # Extract data
        y = df1.iloc[gep1_idx]  # y-axis values
        x = df2.iloc[gep2_idx]  # x-axis values
        
        # Calculate residuals (y - x)
        residuals = y - x
        
        # Create DataFrame with gene names and residuals
        residuals_df = pd.DataFrame({
            'gene': y.index,
            'residual': residuals.values
        })
        
        # Save to CSV
        csv_filename = f"GEP{gep1_idx + 1}_vs_GEP{gep2_idx + 1}_residuals.csv"
        csv_path = os.path.join(output_dir, csv_filename)
        residuals_df.to_csv(csv_path, index=False, sep='\t', header=False)
        
        # Create scatterplot
        ax.scatter(x, y, alpha=0.6, s=30)
        
        # Add diagonal reference line
        min_val = min(x.min(), y.min())
        max_val = max(x.max(), y.max())
        ax.plot([min_val, max_val], [min_val, max_val], 
                'r--', alpha=0.5, linewidth=1, label='y=x')
        
        # Calculate correlation
        correlation = np.corrcoef(x, y)[0, 1]
        
        # Labels and title
        ax.set_xlabel(f"GEP {gep2_idx + 1}", fontsize=10)
        ax.set_ylabel(f"GEP {gep1_idx + 1}", fontsize=10)
        ax.set_title(f"GEP {gep1_idx + 1} vs GEP {gep2_idx + 1}\n(r = {correlation:.3f})", fontsize=11)
        ax.legend(fontsize=8)
        ax.grid(True, alpha=0.3)
    
    # Hide empty subplots
    for idx in range(n_plots, len(axes)):
        axes[idx].set_visible(False)
    
    plt.tight_layout()
    return fig, axes
```

```python
s1_k8 = pd.read_csv('all_datasets_consensus/anal_pc5_c21_S1.filtered_consensus_outputs/anal_pc5_c21_S1.filtered.gene_spectra_score.k_8.dt_0_01.txt', sep='\t', index_col=0)
s1_k8 = s1_k8.dropna(axis='columns', how='any')
s1_k8.head(7)
```

```python
s1_k9 = pd.read_csv('all_datasets_consensus/anal_pc5_c21_S1.filtered_consensus_outputs/anal_pc5_c21_S1.filtered.gene_spectra_score.k_9.dt_0_01.txt', sep='\t', index_col=0)
s1_k9 = s1_k9.dropna(axis='columns', how='any')
s1_k9.head(7)
```

```python
gep_pairs = [
    (1, 1),  # GEP 2 vs GEP 2
    (1, 5),
    (3, 5)
]

fig, axes = plot_gep_comparisons(
    df1=s1_k8,
    df2=s1_k9,
    gep_pairs=gep_pairs,
    figsize=(10, 5),
    ncols=2,
    output_dir='s1_k8_vs_k9_high-corr-geps'
)

plt.savefig('gep_comparison_plots.png', dpi=300, bbox_inches='tight')
plt.show()
```
