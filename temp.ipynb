```python
import anndata as ad

v1 = ad.read_h5ad("patient_data/anal_pc5_c21_S1.filtered.h5ad")
v2 = ad.read_h5ad("patient_data/A10.h5ad")

# High-level overview
print("=== V1 ===")
print(v1)
print("\n=== V2 ===")
print(v2)
```

```python
# Column names in obs and var
print("obs columns V1:", v1.obs.columns.tolist())
print("obs columns V2:", v2.obs.columns.tolist())

print("var columns V1:", v1.var.columns.tolist())
print("var columns V2:", v2.var.columns.tolist())

# Layers
print("layers V1:", list(v1.layers.keys()))
print("layers V2:", list(v2.layers.keys()))

# obsm (embeddings like PCA, UMAP)
print("obsm V1:", list(v1.obsm.keys()))
print("obsm V2:", list(v2.obsm.keys()))

# uns (unstructured metadata)
print("uns V1:", list(v1.uns.keys()))
print("uns V2:", list(v2.uns.keys()))
```

```python
print("X type V1:", type(v1.X))
print("X type V2:", type(v2.X))

# One might be sparse, the other dense
# Or one might be raw counts vs normalized
import numpy as np
print("X max V1:", v1.X.max())
print("X max V2:", v2.X.max())
print("X dtype V1:", v1.X.dtype)
print("X dtype V2:", v2.X.dtype)
```
