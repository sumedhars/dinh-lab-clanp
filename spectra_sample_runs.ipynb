```python
import scanpy as sc
import Spectra as spc
import pandas as pd
from Spectra import Spectra_util as spc_tl
from Spectra import K_est as kst
from Spectra import default_gene_sets
```

```python
annotations = spc.default_gene_sets.load()
```

```python
# Load your data
adata = sc.read_h5ad('patient_data/anal_pc5_c21_S1.filtered.h5ad')
```

```python
metadata = pd.read_csv('data/anal_cancer_n21_meta.csv')
metadata.head()
```

```python
metadata = metadata.set_index('Unnamed: 0')
adata.obs['cluster_annot3'] = metadata.loc[adata.obs_names, 'cluster_annot3']
```

```python
print(adata)
```

```python
print(adata.obs.head())
```

spectra model

```python
model = spc.est_spectra(
    adata=adata,
    gene_set_dictionary=annotations,
    cell_type_key='cluster_annot3',  
    use_highly_variable=True,
    use_weights=True,
    lam=0.1,
    delta=0.001,
    kappa=0.00001,
    rho=0.00001,
    use_cell_types=True,
    n_top_vals=25,
    label_factors=True,
    overlap_threshold=0.2,
    num_epochs=10000
)
```
